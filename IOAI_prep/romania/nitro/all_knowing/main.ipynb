```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

x = np.array(pd.read_csv('train_data.csv').drop(['subtaskID', 'datapointID'], axis=1))
x
```

```python
tries = 500000
best_error = np.inf
best_ids = []
n = 420

for i in range(tries):
    points = x[np.random.choice(n, 4)]
    mean = np.mean(points, axis=0)
    basis = points - mean

    U, S, V = np.linalg.svd(basis, full_matrices=False)
    basis = V[:3]
    x_norm = x - mean
    proj = (x_norm @ basis.T) @ basis
    dists = np.linalg.norm(x_norm - proj, axis=1)
    ids = np.argpartition(dists, 20)[:20]
    err = np.sum(dists[ids]**2)
    if err < best_error:
        print(f"Found better with loss {err:.2f}")
        best_error = err
        best_ids = ids

best_ids
```
