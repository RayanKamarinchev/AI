```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('train_data.csv').drop('datapointID', axis=1)
task1 = df[df['subtaskID'] == 1]
task2 = df[df['subtaskID'] == 2]
task3 = df[df['subtaskID'] == 3]
```

```python
x = task1[['X', 'Y']].to_numpy()
y = task1['answer'].to_numpy()
plt.scatter(task1['X'], task1['Y'], c=task1['answer'])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

model = SVC(kernel='rbf', C=1.0, gamma="scale")
model.fit(x, y)

xx, yy = np.meshgrid(
    np.linspace(x[:, 0].min()-1, x[:, 0].max()+1, 300),
    np.linspace(x[:, 1].min()-1, x[:, 1].max()+1, 300)
)

Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(x[:, 0], x[:, 1], c=y, edgecolors="k")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

tree = DecisionTreeClassifier(max_depth=3)
tree.fit(x, y)

plot_tree(tree)
plt.show()
```

```python
Z = tree.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(x[:, 0], x[:, 1], c=y, edgecolors="k")
plt.show()
```
