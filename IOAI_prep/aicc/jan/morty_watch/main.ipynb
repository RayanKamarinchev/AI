```python
import kagglehub

# Download latest version
path = kagglehub.competition_download('mortys-time-paradox-aicc-round-3')

print("Path to competition files:", path)
```

```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
```

```python
trn = pd.read_csv(path + "/train.csv")
tst = pd.read_csv(path + "/test.csv")

x_train = trn.iloc[:900, 1:9].to_numpy()
y_train = trn.iloc[:900, 9:14].to_numpy()
x_val = trn.iloc[900:, 1:9].to_numpy()
y_val = trn.iloc[900:, 9:14].to_numpy()
X_test = tst.iloc[:, 1:9].to_numpy()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
import optuna

def objective(trial):
    n_est = trial.suggest_int('n_est', 5, 100, log=True)

    model = XGBClassifier(n_estimators=n_est)
    model.fit(x_train, y_train)

    preds = model.predict(x_val)
    return accuracy_score(y_val, preds)

study = optuna.create_study(direction='maximize')
study.optimize(objective, 10)
```
